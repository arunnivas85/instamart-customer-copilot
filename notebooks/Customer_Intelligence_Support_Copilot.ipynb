````python

from pathlib import Path
import sqlite3

import numpy as np
import pandas as pd


# ROOT = Path.cwd();
ROOT = Path(r"C:\Arun\AI_Arun\AI_ML\Capstone Project\instamart-customer-copilot")
RAW = ROOT/"data"/"raw"
CLEAN = ROOT/"data"/"clean"
DB = ROOT/"data"/"instamart.db"


def audit():
    for name in ("customers", "products", "orders"):
        df = pd.read_csv(RAW / f"{name}.csv", keep_default_na=False)
        print(f"\n{name}: rows={len(df)}, columns={len(df.columns)}")
        print(df.dtypes)

    orders = pd.read_csv(RAW / "orders.csv", keep_default_na=False)
    print(
        "\nRaw orders:",
        {
            "rows": len(orders),
            "exact_duplicate_rows": int(orders.duplicated().sum()),
            "pending_rows": int(orders["status"].eq("pending").sum()),
            "distinct_order_ids": int(orders["order_id"].nunique()),
            "unconvertible_order_amounts": int(
                pd.to_numeric(orders["order_amount"], errors="coerce").isna().sum()
            ),
            "negative_delivery_minutes": int(
                pd.to_numeric(orders["delivery_minutes"], errors="coerce").lt(0).sum()
            ),
            "blank_delivery_minutes": int(
                orders["delivery_minutes"].astype(str).str.strip().eq("").sum()
            ),
        },
    )

    customers = pd.read_csv(RAW / "customers.csv", keep_default_na=False)
    ages = pd.to_numeric(customers["age"], errors="coerce")
    nonblank_gender = customers.loc[
        customers["gender"].astype(str).str.strip().ne(""), "gender"
    ]
    print(
        "Raw customers:",
        {
            "blank_age": int(customers["age"].astype(str).str.strip().eq("").sum()),
            "age_outside_10_100": int(((ages < 10) | (ages > 100)).sum()),
            "distinct_raw_city_spellings": int(customers["city"].nunique()),
            "distinct_nonblank_gender_spellings": int(nonblank_gender.nunique()),
            "blank_gender": int(
                customers["gender"].astype(str).str.strip().eq("").sum()
            ),
        },
    )


def clean_data():
    CLEAN.mkdir(parents=True, exist_ok=True)

    customers = pd.read_csv(RAW / "customers.csv", keep_default_na=False)
    products = pd.read_csv(RAW / "products.csv")
    orders = pd.read_csv(RAW / "orders.csv", keep_default_na=False)

    # Standardize customer fields.
    customers["city"] = customers["city"].astype(str).str.strip().str.title()
    customers["gender"] = (
        customers["gender"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map({"male": "male", "m": "male", "female": "female", "f": "female"})
        .fillna("unknown")
    )
    customers["age"] = pd.to_numeric(customers["age"], errors="coerce")
    customers.loc[~customers["age"].between(10, 100), "age"] = np.nan

    city_median_age = customers.groupby("city")["age"].transform("median")
    customers["age"] = customers["age"].fillna(city_median_age)
    if customers["age"].isna().any():
        cities = customers.loc[customers["age"].isna(), "city"].unique().tolist()
        raise ValueError(f"Cannot fill ages; no valid median for these cities: {cities}")

    customers["signup_date"] = pd.to_datetime(customers["signup_date"])

    # Deduplicate orders and keep each order's latest record.
    orders["order_ts"] = pd.to_datetime(orders["order_ts"])
    orders = orders.drop_duplicates()
    orders = (
        orders.sort_values(["order_id", "order_ts"], kind="stable")
        .drop_duplicates(subset="order_id", keep="last")
        .copy()
    )

    # Convert order fields to numeric.
    orders["quantity"] = pd.to_numeric(orders["quantity"], errors="coerce")
    orders["discount_pct"] = (
        pd.to_numeric(orders["discount_pct"], errors="coerce").fillna(0)
    )
    orders["order_amount"] = pd.to_numeric(
        orders["order_amount"], errors="coerce"
    )
    orders["delivery_minutes"] = pd.to_numeric(
        orders["delivery_minutes"], errors="coerce"
    )

    # Recalculate missing order amounts from product price.
    price_by_product = products.set_index("product_id")["price_inr"]
    missing_amount = orders["order_amount"].isna()
    orders.loc[missing_amount, "order_amount"] = (
        orders.loc[missing_amount, "product_id"].map(price_by_product)
        * orders.loc[missing_amount, "quantity"]
        * (1 - orders.loc[missing_amount, "discount_pct"] / 100)
    ).round(2)

    # Negative delivery times are invalid. Keep cancelled orders missing.
    orders.loc[orders["delivery_minutes"] < 0, "delivery_minutes"] = np.nan
    cancelled = orders["status"].eq("cancelled")
    orders.loc[cancelled, "delivery_minutes"] = np.nan

    # Fill missing delivery times only for completed orders, using the
    # completed-order median for each product category.
    orders["category"] = orders["product_id"].map(
        products.set_index("product_id")["category"]
    )
    completed = orders["status"].eq("completed")
    completed_orders = orders.loc[completed]
    category_median = completed_orders.groupby("category")[
        "delivery_minutes"
    ].transform("median")
    orders.loc[completed, "delivery_minutes"] = (
        completed_orders["delivery_minutes"].fillna(category_median)
    )
    orders = orders.drop(columns="category")

    # Save cleaned CSVs and load the same data into SQLite.
    dataframes = {
        "customers": customers,
        "products": products,
        "orders": orders,
    }
    for name, df in dataframes.items():
        df.to_csv(CLEAN / f"{name}.csv", index=False)

    DB.parent.mkdir(parents=True, exist_ok=True)
    with sqlite3.connect(DB) as connection:
        for name, df in dataframes.items():
            df.to_sql(name, connection, if_exists="replace", index=False)

    return dataframes


def validate(data):
    customers = data["customers"]
    orders = data["orders"]

    assert customers["city"].nunique() == 6
    assert set(customers["gender"].unique()) <= {"male", "female", "unknown"}
    assert customers["age"].notna().all()
    assert customers["age"].between(18, 60).all()
    assert pd.api.types.is_datetime64_any_dtype(customers["signup_date"])

    assert len(orders) == 12_309
    assert orders["order_id"].is_unique
    assert orders["status"].value_counts().to_dict() == {
        "completed": 11_110,
        "cancelled": 1_199,
    }
    assert pd.api.types.is_datetime64_any_dtype(orders["order_ts"])
    assert orders["order_amount"].notna().all()

    completed = orders["status"].eq("completed")
    cancelled = orders["status"].eq("cancelled")
    assert orders.loc[completed, "delivery_minutes"].notna().all()
    assert orders.loc[cancelled, "delivery_minutes"].isna().all()
    assert round(orders.loc[completed, "order_amount"].sum(), 2) == 2_998_668.80



    completed = orders["status"].eq("completed")
    cancelled = orders["status"].eq("cancelled")

    print(
        f"*Reference after cleaning:* {len(orders):,} rows "
        f"(one per order_id); "
        f"{completed.sum():,} completed and {cancelled.sum():,} cancelled; "
        f"{orders['status'].eq('pending').sum():,} pending rows; "
        f"{orders['order_amount'].isna().sum():,} missing order_amount; "
        f"{orders.loc[completed, 'delivery_minutes'].isna().sum():,} missing "
        f"delivery_minutes among completed orders and "
        f"{orders.loc[cancelled, 'delivery_minutes'].isna().sum():,} missing "
        f"among cancelled orders; the sum of order_amount over completed orders "
        f"is **INR {orders.loc[completed, 'order_amount'].sum():,.2f}**."
    )

    print("\nCustomer counts by city:")
    print(customers["city"].value_counts().to_string())
    print("\nCustomer counts by gender:")
    print(customers["gender"].value_counts().to_string())
    print("\nOrder status counts:")
    print(orders["status"].value_counts().to_string())
    print(
        "\nCompleted order amount total: INR "
        f"{orders.loc[completed, 'order_amount'].sum():,.2f}"
    )
    print("ALL CHECKPOINTS PASSED")
    

    
if __name__ == "__main__":
    audit();
    cleaned_data = clean_data();
    validate(cleaned_data);
````
